### Problem 15

Apply Gram-Schmidt to $$\mathbf{v}_1 = \begin{bmatrix} 1 \\ 1 \end{bmatrix}, \mathbf{v}_2 = \begin{bmatrix} 1 \\ 0 \end{bmatrix}$$ processing the vectors in the given order.
Set `p15` to the second orthonormal vector, choosing its first coordinate to be positive.

```python
# YOUR CHANGES HERE
v1 = np.array([1, 1], dtype=float)
v2 = np.array([1, 0], dtype=float)

# Normalize v1
q1 = v1 / np.linalg.norm(v1)

# Remove q1's component from v2
u2 = v2 - np.dot(v2, q1) * q1

# Normalize u2
q2 = u2 / np.linalg.norm(u2)

# The problem asks for the second orthonormal vector
p15 = q2
```
